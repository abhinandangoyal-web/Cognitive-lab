```python
print("First three elements:", L[:3])
print("Last three elements:", L[-3:])
```

```python
average = sum(L) / len(L)
```

```python
greater_than_average = [x for x in L if x > average]

print("Average:", average)
print("Elements greater than average:", greater_than_average)
```

```python
scores = tuple(L[:8])

print("Scores:", scores)
```

```python
highest = max(scores)
highest_index = scores.index(highest)
lowest = min(scores)
lowest_count = scores.count(lowest)
print("Highest score:", highest)
print("Index of highest score:", highest_index)
print("Lowest score:", lowest)
print("Number of times lowest score appears:", lowest_count)
```

```python
reversed_scores = list(reversed(scores))

print("Reversed tuple as list:", reversed_scores)
```

```python
search_score = int(input("Enter a score to search: "))

if search_score in scores:
    print("First occurrence index:", scores.index(search_score))
else:
    print("Score not present in the tuple.")
```

```python
import random
```

```python
random.seed(1024170333)
```

```python
numbers = [random.randint(100, 900) for _ in range(100)]
print("Random numbers:")
print(numbers)
```

```python
odd_numbers = [x for x in numbers if x % 2 != 0]
print("\nNumber of odd numbers:", len(odd_numbers))
print("Odd numbers:", odd_numbers)
```

```python
even_numbers = [x for x in numbers if x % 2 == 0]

print("\nNumber of even numbers:", len(even_numbers))
print("Even numbers:", even_numbers)
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
            return True
```

```python
prime_numbers = [x for x in numbers if is_prime(x)]

print("\nNumber of prime numbers:", len(prime_numbers))
print("Prime numbers:", prime_numbers)
```

```python
most_frequent = max(set(numbers), key=numbers.count)
frequency = numbers.count(most_frequent)

print("\nMost frequent number:", most_frequent)
print("Number of occurrences:", frequency)
```

```python
digits = [1, 0, 2, 4, 1, 7, 0, 3]
```

```python
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}
print("Set A:", A)
print("Set B:", B)
```

```python
union = A.union(B)
print("\nUnion of A and B:", union)
```

```python
intersection = A.intersection(B)
print("Intersection of A and B:", intersection)
```

```python
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)
print("A - B=", A_minus_B)
print("B - A=", B_minus_A)
```

```python
symmetric_diff = A.symmetric_difference(B)
print("Symmetric Difference =", symmetric_diff)
```

```python
print("Is A a subset of B is", A.issubset(B))
print("Is B a superset of A is", B.issuperset(A))
```

```python
X = int(input("\nEnter a value to remove from set A: "))
A.discard(X)
print("Set A after discard:", A)
```

```python
my_dict = {
    "name": "Abhinandan",
    "roll_no": "1024170176",
    "branch": "CSE",
    "age": 21,
    "city": "Patiala"
}
print("Original dictionary:")
print(my_dict)
```

```python
my_dict["location"] = my_dict.pop("city")
```

```python
my_dict["cgpa"] = 8.11
print("\nAfter renaming city and adding CGPA:")
print(my_dict)
```

```python
my_dict["age"] += 1
print("\nAfter increasing age by 1:")
print(my_dict)
```

```python
dict_pop = my_dict.copy()
removed_branch = dict_pop.pop("branch")
print("\nUsing pop():")
print("Removed value:", removed_branch)
print("Dictionary:", dict_pop)
```

```python
dict_del = my_dict.copy()
del dict_del["branch"]
print("\nUsing del:")
print("Dictionary:", dict_del)
```

```python
print("\nKey-value pairs:")
for key, value in my_dict.items():
    print(f"{key} → {value}")
```

```python
if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail key is not present in the dictionary.")
```
