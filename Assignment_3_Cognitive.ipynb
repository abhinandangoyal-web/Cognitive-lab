```python
import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 36],
    "Salary": [50000, 70000, 65000, 55000, 68000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

df = pd.DataFrame(data)
```

```python
Shape
print("Shape:")
print(df.shape)
```

```python
Summary including data types and non-null counts
print("\nInfo:")
df.info()
```

```python
# Descriptive statistics
print(df.describe())
```

```python
# First 5 rows and last 3 rows

print(df.head())

print(df.tail(3))
```

```python
# Calculations

print("\nAverage Salary:")
print(df["Salary"].mean())

print("\nTotal Bonus:")
print(df["Bonus"].sum())

print("\nYoungest Employee Age:")
print(df["Age"].min())

print("\nHighest Performance Rating:")
print(df["Rating"].max())
```

```python
# Sort by Salary in descending order

df = df.sort_values(by="Salary", ascending=False)

print(df)
```

```python
# Add Performance Category

def category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

df["Performance_Category"] = df["Rating"].apply(category)

print(df)
```

```python
# Identify missing values

print(df.isnull().sum())
```

```python
# Rename Employee_ID to ID

df = df.rename(columns={"Employee_ID": "ID"})

print(df)
```

```python
# Employees with more than 5 years of experience

print(df[df["Years_of_Experience"] > 5])
```

```python
# Employees belonging to IT department

print(df[df["Department"] == "IT"])
```

```python
# Add Tax column

df["Tax"] = df["Salary"] * 0.10

print(df)
```

```python
# Save modified DataFrame

df.to_csv("employees_modified.csv", index=False)

print("File saved successfully!")
```
